## **Portfolio Simulator & Strategy Analysis**

**Goal:**  
Construct optimal portfolios, compare weighting strategies, 
and evaluate a momentum strategy against SPY benchmark.

**Key Questions**  
1. What is the optimal weight allocation across these stocks?
2. Does momentum strategy beat a passive SPY investment?
3. How does equal-weight compare to momentum-weight and market-cap weight?

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.ticker as mtick
import yfinance as yf

TICKERS = ['AAPL', 'MSFT', 'NVDA', 'META', 'GOOGL', 'SPY']
START   = '2020-01-01'
END     = '2025-12-31'
RISK_FREE_RATE = 0.045

prices = pd.read_csv('data/prices_clean.csv', index_col=0, parse_dates=True)
returns = prices.pct_change().dropna()

# Exclude SPY from portfolio stocks
STOCKS  = [t for t in TICKERS if t != 'SPY']
stock_returns = returns[STOCKS]
spy_returns   = returns['SPY']

print(f"Shape: {returns.shape}")
print(f"Date range: {returns.index[0].date()} → {returns.index[-1].date()}")
```

### **Part 1 - Portfolio Metrics Funtion**
Given the input weight of each stock, compute the annualized return, volatility, and sharpe ratio of the portfolio.

```python
import ipywidgets as widgets
from IPython.display import display, clear_output

def portfolio_metrics(weights, returns, risk_free_rate=RISK_FREE_RATE):
    weights = np.array(weights)
    port_daily = returns @ weights
    ann_return = (1 + port_daily).prod() ** (252 / len(port_daily)) - 1
    ann_vol = port_daily.std() * np.sqrt(252)
    sharpe = (ann_return - risk_free_rate) / ann_vol
    return ann_return, ann_vol, sharpe
    
sliders = {ticker: widgets.FloatText(
                value=round(100/len(STOCKS), 1),
                description=f'{ticker} %:',
                step=0.1,
                layout=widgets.Layout(width='200px')
           ) for ticker in STOCKS}

output = widgets.Output()
button = widgets.Button(description='Calculate')

def on_calculate(b):
    with output:
        clear_output()
        try:
            weights_pct = {t: sliders[t].value for t in STOCKS}
            total = sum(weights_pct.values())
            
            # Validation
            if abs(total - 100) > 0.01:
                print(f"Weights sum to {total:.1f}% — must equal 100%. Please adjust.")
                return
            
            if any(v < 0 for v in weights_pct.values()):
                print("Weights cannot be negative.")
                return
            
            # Calculate
            weights = np.array([weights_pct[t] / 100 for t in STOCKS])
            ret, vol, sharpe = portfolio_metrics(weights, stock_returns)
            
            print("=" * 40)
            print(f"Ann. Return:     {ret:.2%}")
            print(f"Ann. Volatility: {vol:.2%}")
            print(f"Sharpe Ratio:    {sharpe:.4f}")
            print("=" * 40)
            
        except Exception as e:
            print(f"Error: {e}")

button.on_click(on_calculate)

# Display
print("Enter portfolio weights (must sum to 100%):")
display(*sliders.values())
display(button)
display(output)
```

### **Part 2 - Efficient Frontier generated with Monte Carlo Simulator**
 - Randomly generate 10,000 weight combinations of the five stocks. For each one, compute that portfolio's annualized return, volatility, and Sharpe ratio.  
 - Look up the max-Sharpe portfolio and min-volatility portfolio among 10,000 results.  
 - The upper edge of all points is efficient frontier.

```python
N_SIMULATIONS = 10000
np.random.seed(42)

results = np.zeros((N_SIMULATIONS, 3 + len(STOCKS)))

for i in range(N_SIMULATIONS):
    w = np.random.dirichlet(np.ones(len(STOCKS)))
    ret, vol, sharpe = portfolio_metrics(w, stock_returns)
    results[i, :3]  = [ret, vol, sharpe]
    results[i, 3:]  = w

cols = ['Return', 'Volatility', 'Sharpe'] + STOCKS
sim_df = pd.DataFrame(results, columns=cols)

max_sharpe_idx  = sim_df['Sharpe'].idxmax()
min_vol_idx     = sim_df['Volatility'].idxmin()
max_sharpe_port = sim_df.iloc[max_sharpe_idx]
min_vol_port    = sim_df.iloc[min_vol_idx]

print("Max Sharpe Portfolio:")
for s in STOCKS:
    print(f"  {s}: {max_sharpe_port[s]:.2%}")
print(f"  Return: {max_sharpe_port['Return']:.2%}  "
      f"Vol: {max_sharpe_port['Volatility']:.2%}  "
      f"Sharpe: {max_sharpe_port['Sharpe']:.4f}")

print("\nMin Volatility Portfolio:")
for s in STOCKS:
    print(f"  {s}: {min_vol_port[s]:.2%}")
print(f"  Return: {min_vol_port['Return']:.2%}  "
      f"Vol: {min_vol_port['Volatility']:.2%}  "
      f"Sharpe: {min_vol_port['Sharpe']:.4f}")

fig, ax = plt.subplots(figsize=(10, 6))

# All simulated portfolios
scatter = ax.scatter(
    sim_df['Volatility'],
    sim_df['Return'],
    c=sim_df['Sharpe'],
    cmap='RdYlGn',
    alpha=0.4,
    s=3
)
plt.colorbar(scatter, ax=ax, label='Sharpe Ratio')

# Max Sharpe
ax.scatter(max_sharpe_port['Volatility'], max_sharpe_port['Return'],
           color='gold', s=250, zorder=5, marker='*',
           label=f"Max Sharpe ({max_sharpe_port['Sharpe']:.2f})",
           edgecolors='black', linewidth=0.8)

# Min Vol
ax.scatter(min_vol_port['Volatility'], min_vol_port['Return'],
           color='cyan', s=150, zorder=5, marker='*',
           label=f"Min Vol ({min_vol_port['Volatility']:.2%})",
           edgecolors='black', linewidth=0.8)

# Individual stocks
for ticker in STOCKS:
    r = stock_returns[ticker].mean() * 252
    v = stock_returns[ticker].std() * np.sqrt(252)
    ax.scatter(v, r, s=80, zorder=6, edgecolors='white', linewidth=0.8)
    ax.annotate(ticker, (v, r), textcoords='offset points',
                xytext=(6, 4), fontsize=9, color='white',
                bbox=dict(boxstyle='round,pad=0.2', facecolor='black', alpha=0.5))

ax.set_title('Efficient Frontier (10,000 Monte Carlo Simulations)', fontsize=13)
ax.set_xlabel('Annualized Volatility')
ax.set_ylabel('Annualized Return')
ax.xaxis.set_major_formatter(mtick.PercentFormatter(xmax=1))
ax.yaxis.set_major_formatter(mtick.PercentFormatter(xmax=1))
ax.legend(fontsize=9)
ax.grid(alpha=0.3)

plt.tight_layout()
plt.savefig('charts/efficient_frontier.png', dpi=150)
plt.show()
```

### **Part 3 - Strategy Comparison**
Compare three weighting strategies:
1. **Equal-weight** — 1/N for each stock
2. **Momentum-weight** — give more weight to stocks that went up the most over the past 12 months. Stocks that went down get zero weight
3. **Volatility-inverse weight** — weight inversely proportional to volatility 
   (lower risk stocks get higher weight)

Rebalance monthly, compare cumulative returns vs SPY.

```python
def get_monthly_weights(strategy, returns_window):
    n = len(returns_window.columns)

    if strategy == 'equal':
        return np.array([1/n] * n)

    elif strategy == 'momentum':
        cum_returns = (1 + returns_window).prod() - 1
        cum_returns = cum_returns.clip(lower=0)
        if cum_returns.sum() == 0:
            return np.array([1/n] * n)
        return (cum_returns / cum_returns.sum()).values

    elif strategy == 'vol_inverse':
        vols    = returns_window.std()
        inv_vol = 1 / vols
        return (inv_vol / inv_vol.sum()).values


def backtest_strategy(strategy, returns, lookback=252):
    portfolio_returns = []
    dates             = []
    rebal_dates       = returns.resample('ME').last().index
    current_weights   = np.array([1/len(returns.columns)] * len(returns.columns))

    for i in range(lookback, len(returns)):
        date = returns.index[i]
        if date in rebal_dates:
            window          = returns.iloc[i-lookback:i]
            current_weights = get_monthly_weights(strategy, window)
        daily_ret = np.dot(current_weights, returns.iloc[i])
        portfolio_returns.append(daily_ret)
        dates.append(date)

    return pd.Series(portfolio_returns, index=dates, name=strategy)


print("Running backtests...")
equal_ret    = backtest_strategy('equal',       stock_returns)
momentum_ret = backtest_strategy('momentum',    stock_returns)
volinv_ret   = backtest_strategy('vol_inverse', stock_returns)
spy_aligned  = spy_returns.reindex(equal_ret.index)
print("Done.")
```

```python
def strategy_stats(returns_series, name, risk_free=RISK_FREE_RATE):
    ann_ret  = returns_series.mean() * 252
    ann_vol  = returns_series.std() * np.sqrt(252)
    sharpe   = (ann_ret - risk_free) / ann_vol
    cum      = (1 + returns_series).cumprod()
    max_dd   = ((cum - cum.cummax()) / cum.cummax()).min()
    win_rate = (returns_series > 0).mean()

    return {
        'Strategy':     name,
        'Ann. Return':  f"{ann_ret:.2%}",
        'Ann. Vol':     f"{ann_vol:.2%}",
        'Sharpe':       f"{sharpe:.4f}",
        'Max Drawdown': f"{max_dd:.2%}",
        'Win Rate':     f"{win_rate:.2%}"
    }

stats_table = pd.DataFrame([
    strategy_stats(equal_ret,    'Equal Weight'),
    strategy_stats(momentum_ret, 'Momentum Weight'),
    strategy_stats(volinv_ret,   'Vol-Inverse Weight'),
    strategy_stats(spy_aligned,  'SPY Benchmark'),
])

print(stats_table.to_string(index=False))
```

```python
strategies = {
    'Equal Weight':       equal_ret,
    'Momentum Weight':    momentum_ret,
    'Vol-Inverse Weight': volinv_ret,
    'SPY Benchmark':      spy_aligned
}
colors = ['steelblue', 'gold', 'mediumseagreen', 'tomato']

fig, axes = plt.subplots(2, 1, figsize=(12, 10))

# Top: Cumulative returns
ax = axes[0]
for (name, rets), color in zip(strategies.items(), colors):
    cum = (1 + rets).cumprod()
    lw  = 2.2 if name == 'SPY Benchmark' else 1.5
    ls  = '--' if name == 'SPY Benchmark' else '-'
    ax.plot(cum.index, cum.values, label=name,
            color=color, linewidth=lw, linestyle=ls)

ax.set_title('Cumulative Returns by Strategy vs SPY (Monthly Rebalancing)', fontsize=12)
ax.set_ylabel('Growth of $1')
ax.legend(fontsize=9)
ax.grid(alpha=0.3)

# Bottom: Drawdown
ax = axes[1]
for (name, rets), color in zip(strategies.items(), colors):
    cum      = (1 + rets).cumprod()
    drawdown = (cum - cum.cummax()) / cum.cummax()
    lw  = 2.2 if name == 'SPY Benchmark' else 1.5
    ls  = '--' if name == 'SPY Benchmark' else '-'
    ax.plot(drawdown.index, drawdown.values, label=name,
            color=color, linewidth=lw, linestyle=ls)
    ax.fill_between(drawdown.index, drawdown.values, 0,
                    alpha=0.08, color=color)

ax.set_title('Drawdown by Strategy', fontsize=12)
ax.set_ylabel('Drawdown')
ax.yaxis.set_major_formatter(mtick.PercentFormatter(xmax=1))
ax.legend(fontsize=9)
ax.grid(alpha=0.3)

plt.tight_layout()
plt.savefig('charts/strategy_comparison.png', dpi=150)
plt.show()
```

### **Key Observations**

- **Best strategy:** Momentum Weight Strategy achieved the highest Sharpe Ratio of 1.0190. Momentum Strategy outperforms SPY by 23.5% annually, but with higher risk. Momentum Strategy has a volatility of 33.5%, which is almost twice of SPY's volatility 17.1%.
- **Risk Return Tradeoff** Vol-Inverse Max Drawdown is 3.5% higher than Equal Weight because high-volatility stocks(eg, NVDA) receive smaller allocations.
- **Efficient Frontier:** Max Sharpe portfolio allocated 86.48% to NVDA, reflecting dominance of high-return assets in unconstrained optimization.
