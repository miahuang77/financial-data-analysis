# **Financial Data Acquisition**

**Goal**  
 Download price data for 5 tech stocks and SPY, and clean data to get prepared for further study.

**Stocks**  
AAPL,MSFT, NVDA, META, GOOGL, SPY

**Period**  
2020-01-01 to 2025-12-31

```python
#Import
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib.gridspec as dridspec
try:
    import yfinance as yf
except ImportError:
    raise ImportError("Please install yfinance first")
#Settings
TICKERS = ['AAPL', 'MSFT', 'NVDA', 'META', 'GOOGL', 'SPY']
START   = '2020-01-01'
END     = '2025-12-31'
```

### **Download Data**

```python
raw = yf.download(TICKERS, start=START, end=END)

prices = raw['Close']
prices = prices.dropna()

def data_quality_report(df: pd.DataFrame) -> None:
    print("===== Data Quality Report =====")
    print(f"Shape: {df.shape[0]} rows x {df.shape[1]} assets")
    print(f"Date range: {df.index.min().date()} to {df.index.max().date()}")
    print(f"Missing values: {int(df.isna().sum().sum())}")
    print(f"Duplicate dates: {int(df.index.duplicated().sum())}")
    print("\nMissing values by asset:")
    print(df.isna().sum())

data_quality_report(prices)
```

```python
import os
os.makedirs('data', exist_ok=True)
prices.to_csv('data/prices_clean.csv')
```
